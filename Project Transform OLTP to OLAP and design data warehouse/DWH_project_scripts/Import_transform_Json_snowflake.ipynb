## Transform File Json Person-GeneralContact

```python
import json
import pandas as pd
from typing import List, Dict, Any
json_path_1 = "Person-GeneralContact.json"

# ────────────────────────────────────────────────────────────────────────────────
# Helper lambda: lấy phần tử đầu tiên hoặc trả "" nếu không tồn tại
first_or_empty = lambda lst: (lst or [""])[0] if isinstance(lst, list) else ""

def parse_person_data(json_path: str) -> pd.DataFrame:
    """
    Đọc file JSON và trả về DataFrame phẳng.
    """
    with open(json_path, "r", encoding="utf-8") as f:
        data: List[Dict[str, Any]] = json.load(f)

    records: List[Dict[str, Any]] = []

    for item in data:
        person        = first_or_empty(item.get("PersonInfo"))
        contact       = first_or_empty(person.get("PersonContact", []))
        address       = first_or_empty(contact.get("PersonAddress", []))
        address_detail = first_or_empty(address.get("PersonAddressDetail", []))

        records.append({
            "BusinessEntityID"   : item.get("BusinessEntityID"),
            "PersonType"         : item.get("PersonType"),
            "ModifiedDate"       : item.get("ModifiedDate"),
            "FirstName"          : person.get("FirstName", ""),
            "MiddleName"         : person.get("MiddleName", ""),
            "LastName"           : person.get("LastName", ""),
            "PersonAddressDetail": address_detail or ""   # rỗng = ""
        })

    return pd.DataFrame(records)
```
